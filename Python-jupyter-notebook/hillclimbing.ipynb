```python
def hill_climbing():

    current = 0

    def fitness(x):
        return -(x**2) + 4*x   # simple function

    while True:

        neighbors = [current-1, current+1]

        next_state = current

        for n in neighbors:

            if fitness(n) > fitness(next_state):
                next_state = n

        if next_state == current:
            return current

        current = next_state


result = hill_climbing()

print("Best state:",result)
```
