```python
#Task 
from collections import deque
import heapq
def BFS(cube, start, goal):
    size, rowLen = len(cube), len(cube[0])
    queue = deque([start])#WILL BE i and j position of start in queue
    visited = set([start])
    parent = {
        start: None
    }
    
    while queue:
        x, y = queue.popleft()
        if (x,y) == goal:
            return reconstruct_path(parent, goal)
            
        for nx, ny in get_neighbors(x,y,size,rowLen):
            if cube[nx][ny] == 0 and (nx,ny) not in visited:#if path is 0 and not visited then add in qeuue
                visited.add((nx,ny))
                parent[(nx,ny)] = (x,y)
                queue.append((nx,ny))
    return -1

def get_neighbors(x, y, n, m):
    moves = [(1,0), (-1,0), (0,1), (0,-1)]
    for dx, dy in moves:
        nx= x+dx
        ny = y+dy
        if 0 <= nx < n and 0 <= ny < m:#check boundry
            yield nx, ny#generator 

def reconstruct_path(parent, goal):
    path = []
    node = goal
    while node is not None:
        path.append(node)
        node = parent.get(node)
    return path[::-1]#return path opposite way
    
def DFS(cube, start, goal):
#same logic just use stack in it
    n, m = len(cube), len(cube[0])
    stack = [start]
    visited = set([start])
    parent = {start: None}
    
    while stack:
        x, y = stack.pop()
        if (x,y) == goal:
            return reconstruct_path(parent, goal)
        
        for nx, ny in get_neighbors(x,y,n,m):
            if cube[nx][ny] == 0 and (nx,ny) not in visited:
                visited.add((nx,ny))
                parent[(nx,ny)] = (x,y)
                stack.append((nx,ny))
    return -1
#logic same just use priority queue and cost of path
def UCS(cube, start, goal):
    n, m = len(cube), len(cube[0])
    priorityq = [(0, start)]
    parent = {start: None}
    cost = {start: 0}
    
    while priorityq:
        c, (x,y) = heapq.heappop(priorityq)
        if (x,y) == goal:
            return reconstruct_path(parent, goal)
        
        for nx, ny in get_neighbors(x,y,n,m):
            if cube[nx][ny] == 0:
                new_cost = c + 1
                if (nx,ny) not in cost or new_cost < cost[(nx,ny)]:
                    cost[(nx,ny)] = new_cost
                    parent[(nx,ny)] = (x,y)
                    heapq.heappush(priorityq, (new_cost, (nx,ny)))#lowest cost will go on first 
    return -1
#will go with lomit
def dls(cube, start, goal, limit):
    size, rowLen = len(cube), len(cube[0])
    stack = [(start,0)]
    parent = {start: None}
    visited = set()
    
    while stack:
        (x,y), depth = stack.pop()
        if (x,y) == goal:
            return reconstruct_path(parent, goal)
        if depth < limit:
            for nx, ny in get_neighbors(x,y,size,rowLen):
                if cube[nx][ny] == 0 and (nx,ny) not in visited:
                    visited.add((nx,ny))
                    parent[(nx,ny)] = (x,y)
                    stack.append(((nx,ny), depth+1))
    return None

def ids(cube, start, goal):
    limit = 0
    while True:
        path = dls(cube, start, goal, limit)
        if path:
            return path
        limit += 1
        if limit > len(cube)*len(cube[0]): 
            return -1


def get_cube(filename):
    cube = []
    start, goal = None, None
    with open(filename, 'r') as f:
        for i, line in enumerate(f):
            row = []
            for j, ch in enumerate(line.strip()):
                if ch == 'S':
                    start = (i, j)
                    row.append(0)
                elif ch == 'G':
                    goal = (i, j)
                    row.append(0)
                else:
                    row.append(int(ch))
            cube.append(row)
    return cube, start, goal

cube, start, goal = get_cube("cube.txt")

print("BFS:", BFS(cube, start, goal))
print("DFS:", DFS(cube, start, goal))
print("UCS:", UCS(cube, start, goal))
print("IDS:", ids(cube, start, goal))
```
