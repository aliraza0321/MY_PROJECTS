```python
from collections import deque

graph={
    "A" :[ "B","C"],
"B": ["D","E"],
"C" :["E"],
"D" : [],
"E" :["A"]
}
def bfs(graph,start):
    visited = set()
    queue = deque([start])
    while queue:
        node = queue.popleft()
        if node not in visited:
            print(node)
            visited.add(node)
            queue.extend(graph[node])

bfs(graph,"B")
```
