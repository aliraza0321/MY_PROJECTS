```python
import time
import heapq
from collections import deque

def print_state(state):
    for i in range(0, 9, 3):
        print(" ".join(state[i:i+3]))
    print("-----")
    
def get_nearest(state):
    neartest = []
    idx = state.index("0")
    row, col = divmod(idx, 3)
    moves = [(-1,0),(1,0),(0,-1),(0,1)]
    for dr, dc in moves: # row direction dr column direction dc 
        r, c = row+dr, col+dc
        if 0 <= r < 3 and 0 <= c < 3:
            new_idx = r*3+c
            new_state = list(state)# convert into list for swaping
            new_state[idx], new_state[new_idx] = new_state[new_idx], new_state[idx]
            neartest.append("".join(new_state))# append posi
    return neartest

def dfs(start, goal):
    stack = [(start, [start])]
    visited = []
    while stack:
        state, path = stack.pop()
        if state == goal:
            return path, visited
        if state not in visited:
            visited.append(state)
            for near in get_nearest(state):
                if near not in visited:
                    stack.append((near, path+[near]))
    return None, visited

def bfs(start, goal):
    queue = deque([(start, [start])])
    visited = [start]
    while queue:
        state, path = queue.popleft()
        if state == goal:
            return path, visited
        for near in get_nearest(state):
            if near not in visited:
                visited.append(near)
                queue.append((near, path+[near]))
    return None, visited

def dls(state, goal, depth, visited):
    if state == goal:
        return [state]
    if depth == 0:
        return None
    visited.append(state)
    for near in get_nearest(state):
        if near not in visited:
            path = dls(near, goal, depth-1, visited)
            if path:
                return [state]+path
    return None

def ids(start, goal):
    depth = 0
    while True:
        visited = []
        path = dls(start, goal, depth, visited)
        if path:
            return path, visited
        depth += 1

def ucs(start, goal):
    pq = [(0, start, [start])]
    visited = {}
    while pq:
        cost, state, path = heapq.heappop(pq)
        if state == goal:
            return path, visited
        if state not in visited or cost < visited[state]:
            visited[state] = cost
            for near in get_nearest(state):
                heapq.heappush(pq, (cost+1, near, path+[near]))
    return None, visited

start = input("Enter start State:")
goal = input("Enter goal State:")

#bfs algorithm calling
print("-----------------")
print("BFS Algorithm")
print("-----------------")
start_time = time.time()
path, visited = bfs(start, goal)
end_time = time.time()
if path:
    print(f"Time taken: {end_time-start_time:.6f} seconds")
    print(f"Path Cost: {len(path)-1}")
    print(f"No of Node Visited: {len(visited)}")
    for state in path:
        print_state(state)
else:
    print("No solution found.")

#UCS algorithm calling
print("-----------------")
print("UCS Algorithm")
print("-----------------")
start_time = time.time()
path, visited = ucs(start, goal)
end_time = time.time()
if path:
    print(f"Time taken: {end_time-start_time:.6f} seconds")
    print(f"Path Cost: {len(path)-1}")
    print(f"No of Node Visited: {len(visited)}")
    for state in path:
        print_state(state)
else:
    print("No solution found.")

#Ids algorithm calling
print("-----------------")
print("IDS Algorithm")
print("-----------------")
start_time = time.time()
path, visited = ids(start, goal)
end_time = time.time()
if path:
    print(f"Time taken: {end_time-start_time:.6f} seconds")
    print(f"Path Cost: {len(path)-1}")
    print(f"No of Node Visited: {len(visited)}")
    for state in path:
        print_state(state)
else:
    print("No solution found.")

#DFS algorithm caling

print("-----------------")
print("DFS Algorithm")
print("-----------------")
start_time = time.time()
path, visited = dfs(start, goal)
end_time = time.time()
if path:
    print(f"Time taken: {end_time-start_time:.6f} seconds")
    print(f"Path Cost: {len(path)-1}")
    print(f"No of Node Visited: {len(visited)}")
    for state in path:
        print_state(state)
else:
    print("No solution found.")

```
