```python
from queue import PriorityQueue
graph = {
    "S": ["A","B"],
    "A": ["C"],
    "B": ["D"],
    "C": [],
    "D": ["G"],
    "G": []
}
heuristic = {
    "S":7,
    "A":2,
    "B":4,
    "C":5,
    "D":2,
    "G":0
}
def greedy_search(start,goal):
    nodes=[]
    visited = set()
    pq = PriorityQueue()
    pq.put((heuristic[start],start))#in priority queue lower h(n) values will be on top 
    while not pq.empty():
        h,node = pq.get()
        if node not in visited:
            nodes.append(node)
            visited.add(node)
            if node == goal:
                print("Goal reached")
                return nodes
            for neighbor in graph[node]:
                pq.put((heuristic[neighbor],neighbor))

print(greedy_search("S","G"))
```
