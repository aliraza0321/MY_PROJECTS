```python
import heapq

graph = {
"A": [("B",1),("C",4)],
"B": [("D",2)],
"C": [("D",1)],
"D": []
}
def ucs(graph,start):
    visited = set()
    queue = [(0,start)]
    while queue:
        cost,node = heapq.heappop(queue)
        if node not in visited:
            print(node,"cost:",cost)
            visited.add(node)
            for neighbour,weight in graph[node]:
                heapq.heappush(queue,(cost+weight,neighbour))

ucs(graph,"A")
```

```python
def dfs_limited(graph,node,depth,visited):

    if depth < 0:
        return

    print(node)

    visited.add(node)

    for neighbour in graph[node]:

        if neighbour not in visited:

            dfs_limited(graph,neighbour,depth-1,visited)


def ids(graph,start,max_depth):

    for depth in range(max_depth+1):

        print("Depth:",depth)

        visited=set()

        dfs_limited(graph,start,depth,visited)
```
