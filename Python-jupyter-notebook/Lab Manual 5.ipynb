```python
#Question no 1 for for A* search
from collections import deque
import heapq

def get_neighbors(x, y, n, m):
    moves = [(1,0), (-1,0), (0,1), (0,-1)]
    for dx, dy in moves:
        nx= x+dx
        ny = y+dy
        if 0 <= nx < n and 0 <= ny < m:#check boundry
            yield nx, ny#generator 

def reconstruct_path(parent, goal):
    path = []
    node = goal
    while node is not None:
        path.append(node)
        node = parent.get(node)
    return path[::-1]#return path opposite way

#logic same just use priority queue and cost of path

def UCS(cube, start, goal):
    n, m = len(cube), len(cube[0])

    # heuristic (Manhattan distance)
    def heuristic(x, y):
        return abs(goal[0] - x) + abs(goal[1] - y)
    a = heuristic(start[0], start[1])
    priorityq = [(a, 0, start)]   

    parent = {start: None}
    cost = {start: 0}            
    short_walls = {start: 0}

    while priorityq:
        c, sw_count, (x, y) = heapq.heappop(priorityq)

        if (x, y) == goal:
            return reconstruct_path(parent, goal)

        for nx, ny in get_neighbors(x, y, n, m):

            # High wall cannot pass
            if cube[nx][ny] == 1:
                continue
            # Step cost
            step_cost = 1
            # Count short walls
            new_sw = sw_count
            if cube[nx][ny] == 2:
                new_sw += 1
            new_cost = cost[(x, y)] + step_cost   # g(n)
            if (nx, ny) not in cost or new_cost < cost[(nx, ny)]:
                cost[(nx, ny)] = new_cost
                short_walls[(nx, ny)] = new_sw
                parent[(nx, ny)] = (x, y)
                a = heuristic(nx, ny)
                f = new_cost + a   # f(n) = g(n) + h(n)
                heapq.heappush(priorityq, (f, new_sw, (nx, ny)))
    return -1

def get_cube(filename):
    cube = []
    start, goal = None, None
    with open(filename, 'r') as f:
        for i, line in enumerate(f):
            row = []
            for j, ch in enumerate(line.strip()):
                if ch == 'S':
                    start = (i, j)
                    row.append(0)
                elif ch == 'G':
                    goal = (i, j)
                    row.append(0)
                else:
                    row.append(int(ch))
            cube.append(row)
    return cube, start, goal

cube, start, goal = get_cube("cube.txt")



print("UCS:", UCS(cube, start, goal))

```

```python
#Question 2 Tik tak toe
"""
Tic Tac Toe Player with Minimax Algorithm
"""
import copy
import math

X = "X"
O = "O"
EMPTY = None
def initial_state():
    return [[EMPTY, EMPTY, EMPTY],
            [EMPTY, EMPTY, EMPTY],
            [EMPTY, EMPTY, EMPTY]]


def player(board):
    count = 0
    for i in board:
        for j in i:
            if j is not None:
                count += 1
    return O if count % 2 != 0 else X


def actions(board):
    res = set()
    for i in range(3):
        for j in range(3):
            if board[i][j] == EMPTY:
                res.add((i, j))
    return res


def result(board, action):
    (i, j) = action
    if board[i][j] is not EMPTY:
        raise Exception("Invalid move")

    new_board = copy.deepcopy(board)
    new_board[i][j] = player(board)
    return new_board


def winner(board):
    lines = []

    # Rows
    lines.extend(board)

    # Columns
    for j in range(3):
        lines.append([board[0][j], board[1][j], board[2][j]])

    # Diagonals
    lines.append([board[0][0], board[1][1], board[2][2]])
    lines.append([board[0][2], board[1][1], board[2][0]])

    for line in lines:
        if line[0] is not None and line.count(line[0]) == 3:
            return line[0]

    return None


def terminal(board):
    if winner(board) is not None:
        return True

    for row in board:
        if EMPTY in row:
            return False

    return True


def utility(board):
    win = winner(board)
    if win == X:
        return 1
    elif win == O:
        return -1
    return 0


def minimax(board):

    if terminal(board):
        return None

    turn = player(board)

    if turn == X:
        best_val = -math.inf
        best_move = None
        for action in actions(board):
            val = min_value(result(board, action))
            if val > best_val:
                best_val = val
                best_move = action
        return best_move

    else:
        best_val = math.inf
        best_move = None
        for action in actions(board):
            val = max_value(result(board, action))
            if val < best_val:
                best_val = val
                best_move = action
        return best_move


def max_value(board):
    if terminal(board):
        return utility(board)

    v = -math.inf
    for action in actions(board):
        v = max(v, min_value(result(board, action)))
    return v


def min_value(board):
    if terminal(board):
        return utility(board)

    v = math.inf
    for action in actions(board):
        v = min(v, max_value(result(board, action)))
    return v
```

```python
# main part of tik tak toe
if __name__ == "__main__":

    board = initial_state()
    user = input("Choose player (X or O): ").upper()

    while True:

        print("\nCurrent Board:")
        for row in board:
            print(row)

        if terminal(board):
            win = winner(board)
            if win:
                print(f"\nGame Over! {win} wins!")
            else:
                print("\nGame Over! It's a Draw!")
            break

        turn = player(board)

        if turn == user:
            print("\nYour Turn")
            i = int(input("Row (0-2): "))
            j = int(input("Col (0-2): "))
            board = result(board, (i, j))

        else:
            print("\nAI Turn")
            move = minimax(board)
            board = result(board, move)
```
