```python
def minimax(values, depth, maximizing):
    if depth == 2:
        return values.pop(0)
    if maximizing:
        best = -1000
        for i in range(2):
            val = minimax(values, depth+1, False)
            best = max(best, val)
        return best
    else:
        best = 1000
        for i in range(2):
            val = minimax(values, depth+1, True)
            best = min(best, val)
        return best

values = [7,5,2,9]

result = minimax(values,0,True)

print("Best value:",result)
```
